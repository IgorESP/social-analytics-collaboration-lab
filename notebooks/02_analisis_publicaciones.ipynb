# Análisis de publicaciones

Actividad de los autores, hashtags más utilizados e interacciones recibidas.

> Antes de hacer commit, limpia los outputs (ver `docs/NOTEBOOKS.md`).

```python
from social_analytics.data_loader import load_dataset
from social_analytics.posts import posts_per_user, top_hashtags
```

```python
ds = load_dataset()
df_posts = ds.posts
df_interactions = ds.interactions
print(f"Publicaciones: {len(df_posts)} | Interacciones: {len(df_interactions)}")
```

## Actividad por autor

```python
actividad = posts_per_user(df_posts)
actividad.head(10)
```

## Hashtags más utilizados

```python
hashtags = top_hashtags(df_posts, limit=10)
hashtags
```

```python
ax = hashtags.sort_values().plot.barh(figsize=(6, 4), title="Top 10 hashtags")
ax.set_xlabel("Número de publicaciones");
```

## Interacciones por tipo

```python
por_tipo = df_interactions["kind"].value_counts().sort_index()
por_tipo.plot.bar(title="Interacciones por tipo", rot=0, color="#55A868");
```

## Conclusiones

- Unos pocos autores concentran gran parte de las publicaciones.
- Los likes son, con diferencia, la interacción más frecuente.
