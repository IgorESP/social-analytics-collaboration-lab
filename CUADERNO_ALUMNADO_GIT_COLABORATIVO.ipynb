## 6. Las ramas del laboratorio

```text
main (v1.0.0)                     versión estable: no se trabaja en ella
 └── lab/<id>/base (etiqueta)     punto de partida de cada caso + ESCENARIO.md
      ├── ejercicio/<id>-alumno-a commits ya hechos por la persona A
      ├── ejercicio/<id>-alumno-b commits ya hechos por la persona B
      └── integracion/<id>        destino de las PR del escenario
```

- Las ramas A y B de cada escenario parten **del mismo commit**.
- Las PR **nunca** van a `main`: van a `integracion/<id>`. Así un escenario no afecta a los demás.
- Las etiquetas `lab/<id>/...` guardan el estado original de cada rama. `scripts/lab.py restaurar <id>` permite repetir un
  escenario y guarda antes vuestro trabajo en `respaldo/<id>/...`.
- Los conflictos **no están resueltos de antemano**: aparecen cuando integráis.

## 8. Comandos Git de referencia

| Para… | Comando |
|-------|---------|
| Ver ramas locales y remotas | `git branch -a` |
| Cambiar de rama (la crea desde origin si hace falta) | `git switch <rama>` |
| Crear una rama nueva desde otra | `git switch -c fix/g03-dashboard integracion/g03` |
| Historial en forma de grafo | `git log --oneline --graph --decorate <rama1> <rama2>` |
| Commits de B que no están en A | `git log --oneline A..B` |
| Antepasado común de dos ramas | `git merge-base A B` |
| Diferencias entre ramas (resumen / completo) | `git diff --stat A B` · `git diff A B -- <fichero>` |
| Cambios desde que una rama salió de otra | `git diff A...B` |
| Quién cambió cada línea | `git blame <fichero>` |
| Historia de un fichero aunque se renombre | `git log --follow -p -- <fichero>` |
| Buscar texto en el proyecto | `git grep <texto>` |
| Añadir cambios por partes | `git add -p` |
| Traer cambios del remoto | `git fetch` · `git pull` |
| Publicar una rama nueva | `git push -u origin <rama>` |
| Integrar otra rama en la actual | `git merge <rama>` |
| Ficheros en conflicto | `git status` · `git diff --name-only --diff-filter=U` |
| Elegir una versión completa durante un conflicto | `git checkout --ours <f>` · `git checkout --theirs <f>` |
| Cancelar un merge | `git merge --abort` |
| Rebasar la rama actual sobre otra | `git rebase <rama>` (y `git rebase --abort`) |
| Reescribir el remoto de forma segura | `git push --force-with-lease` |
| Recuperar un fichero de otra rama | `git restore --source <rama> -- <fichero>` |
| Deshacer commits locales sin perder cambios | `git reset --soft <commit>` |
| Guardar cambios temporalmente | `git stash` · `git stash pop` |
| Estado y restauración de escenarios | `uv run python scripts/lab.py escenarios` · `... restaurar <id>` |
| Tests | `uv run pytest` · `uv run pytest -m "not notebook"` |
| Diferencias entre notebooks | `uv run nbdiff A B -- <notebook>` · `uv run nbdiff-web ...` |
